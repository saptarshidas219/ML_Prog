```python
from sklearn import tree
features = [
    [150, 0],  # Apple
    [170, 0],  # Apple
    [130, 1],  # Orange
    [120, 1],  # Orange
    [250,2],[190,2],[230,2]
]
labels = ["apple", "apple", "orange", "orange","orange","watermelon","watermelon"]
clf = tree.DecisionTreeClassifier()
clf = clf.fit(features, labels)
predict = clf.predict([[160, 0],[30,1],[150,1],[170,2],[100,2]])
print(predict)
```

```python
#Plot the decision tree
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt
plt.figure(figsize=(18, 6))
plot_tree(clf,feature_names=["weight","color"],class_names=["apple","orange","watermelon"],filled=True)
plt.show()
```

```python
#Random Forest
from sklearn.ensemble import RandomForestClassifier
features = [
    [150, 0],  # Apple
    [170, 0],  # Apple
    [130, 1],  # Orange
    [120, 1],  # Orange
    [250,2],[190,2],[230,2]
]
labels = ["apple", "apple", "orange", "orange","orange","watermelon","watermelon"]

clf = RandomForestClassifier()
clf = clf.fit(features, labels)
predict = clf.predict([[160, 0],[30,1],[150,1],[170,2],[100,2]])
print(predict)
```
