```python
for col in df.columns:
    if df[col].dtype == 'object':
        df[col]=df[col].fillna(df[col].mode()[0])

    else:
        df[col]=df[col].fillna(df[col].mean())
```

```python
for col in df.columns:
    if df[col].dtype == 'int64' or df[col].dtype == 'float64':
        sns.histplot(df[col], kde=True)
        plt.title(f'Distribution of {col}')
        plt.xlabel(col)
        plt.ylabel('Frequency')
        plt.show()
```

```python
for col in df.columns:
    if df[col].dtype == 'int64' or df[col].dtype == 'float64':
        sns.boxplot(x=df[col])
        plt.title(f'Boxplot of {col}')
        plt.xlabel(col)
        plt.show()
```

```python
q1_bmi = df['bmi'].quantile(0.25)
q3_bmi = df['bmi'].quantile(0.75)
iqr_bmi = q3_bmi - q1_bmi
lm_bmi = q1_bmi -1.5*iqr_bmi
up_bmi = q3_bmi + 1.5*iqr_bmi



q1_pc = df.past_consultations.quantile(0.25)
q3_pc = df.past_consultations.quantile(0.75)
iqr_pc = q3_pc - q1_pc
lm_pc = q1_pc - 1.5 * iqr_pc
up_pc = q3_pc + 1.5 * iqr_pc



q1_he = df.Hospital_expenditure.quantile(0.25)
q3_he = df.Hospital_expenditure.quantile(0.75)
iqr_he = q3_he - q1_he
lm_he = q1_he - 1.5 * iqr_he
up_he = q3_he + 1.5 * iqr_he




q1_sal = df.Anual_Salary.quantile(0.25)
q3_sal = df.Anual_Salary.quantile(0.75)
iqr_sal = q3_sal - q1_sal
lm_sal = q1_sal - 1.5 * iqr_sal
up_sal = q3_sal + 1.5 * iqr_sal


q1_charges = df.charges.quantile(0.25)
q3_charges = df.charges.quantile(0.75)
iqr_charges = q3_charges - q1_charges
lm_charges = q1_charges - 1.5 * iqr_charges
up_charges = q3_charges + 1.5 * iqr_charges







```

```python
df_fill = df[(df['bmi'] >= lm_bmi) & (df['bmi'] <= up_bmi) &
             (df['past_consultations'] >= lm_pc) & (df['past_consultations'] <= up_pc) &
             (df['Hospital_expenditure'] >= lm_he) & (df['Hospital_expenditure'] <= up_he) &
             (df['Anual_Salary'] >= lm_sal) & (df['Anual_Salary'] <= up_sal) &
             (df['charges'] >= lm_charges) & (df['charges'] <= up_charges)]
df_fill_copy = df_fill.copy()
```

```python
for col in df_fill.columns:
    if df_fill[col].dtype == 'int64' or df_fill[col].dtype == 'float64':
        sns.boxplot(x=df_fill[col])
        plt.title(f'Boxplot of {col}')
        plt.xlabel(col)
        plt.show()
```

```python
#Found Multicollinearity between Anual_Salary and Hospital_expenditure. We will drop Anual_Salary as it has less correlation with charges than Hospital_expenditure.
# Varince Inflation Factor has been determined, need to find out r^2 & VIF for each variable to determine which one to drop.
#Minimum value of VIF is 1, and there is no upper limit for VIF. However, a common rule of thumb is that a VIF above 5 or 10 indicates a problematic amount of multicollinearity.
#  In such cases, it may be necessary to remove or combine variables to reduce multicollinearity and improve the stability of the regression model.

sns.heatmap(df_fill.corr(), annot=True, cmap='Purples',annot_kws={"size": 8 },fmt='.1f')
```

```python
df_fill = df_fill.drop(columns=['charges'])
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
```

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor
```

```python
vif_data =pd.DataFrame()
```

```python
vif_data['feature'] = df_fill.columns
print(vif_data)
```

```python
vif_data['factors'] = [variance_inflation_factor(df_fill.values, i) for i in range(len(df_fill.columns))]
```

```python
x = df_fill_copy.loc[:, list(vif_data['feature'])] 
y = df_fill_copy['charges']
x['smoker'] = x['smoker'].map({'no': 0, 'yes': 1})
```

```python
x_train, x_test,y_train,y_test = train_test_split(x,y,train_size=0.8,random_state=0)
```

```python
clf = LinearRegression()
clf.fit(x_train,y_train)
```

```python
prediction = clf.predict(x_test)
prediction
```

```python
error_pred = pd.DataFrame(columns=['Actuals','Prediction'])
```

```python
error_pred['Actuals'] = y_test
error_pred['Prediction'] = prediction
```

```python
error_pred['deviation'] = error_pred['Actuals']-error_pred['Prediction']
```
