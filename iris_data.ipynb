```python
from ucimlrepo import fetch_ucirepo 
  
# fetch dataset 
iris = fetch_ucirepo(id=53) 
  
# data (as pandas dataframes) 
X = iris.data.features 
y = iris.data.targets 
  
# metadata 
print(iris.metadata) 
  
# variable information 
print(iris.variables) 
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(X,y,test_size=0.8,random_state=1)
```

```python
from sklearn.ensemble import RandomForestClassifier
CLF_L = RandomForestClassifier()
CLF_L.fit(x_train,y_train)
```

```python
predict = CLF_L.predict(x_test)
```

```python
import pandas as pd
import numpy as np
test_case = pd.DataFrame()
test_case['Actual'] = y_test
test_case['predict'] = predict
```

```python
test_case
```

```python
count = 0
# for i in range(0,len(test_case)):
#     if test_case['Actual'].iloc[i]==test_case['predict'].iloc[i]:

for actual, predict in zip(test_case['Actual'], test_case['predict']):
    if actual == predict:
        count = count +1

accuracy = (count/len(test_case)) * 100
print(accuracy)
```

```python
from sklearn.gaussian_process import GaussianProcessClassifier
CLF_L = GaussianProcessClassifier()
CLF_L.fit(x_train,y_train)
```

```python
predict_2 = CLF_L.predict(x_test)
test_case2 = pd.DataFrame()
test_case2['Actual'] = y_test
test_case2['predict'] = predict_2
```
